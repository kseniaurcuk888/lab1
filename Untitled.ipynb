```python
# Мода для Cabin (категориальный признак)
cabin_mode = df['Cabin'].mode()[0]
df['Cabin'].fillna(cabin_mode, inplace=True)

# Медиана для Age (числовой, устойчива к выбросам)
age_median = df['Age'].median()
df['Age'].fillna(age_median, inplace=True)

# Мода для Embarked (категориальный признак)
embarked_mode = df['Embarked'].mode()[0]
df['Embarked'].fillna(embarked_mode, inplace=True)
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

scaler = MinMaxScaler()
df['Age'] = scaler.fit_transform(df[['Age']])
```

```python
df['Age'].describe()
```

```python
scaler = MinMaxScaler()
df[['Pclass', 'SibSp', 'Parch', 'Fare']] = scaler.fit_transform(df[['Pclass', 'SibSp', 'Parch', 'Fare']])
```

```python
df[['Pclass', 'SibSp', 'Parch', 'Fare']].describe()
```

```python
df = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)
```

```python
df = df.astype({'Sex_male': 'int', 'Embarked_Q': 'int', 'Embarked_S': 'int'})
```
